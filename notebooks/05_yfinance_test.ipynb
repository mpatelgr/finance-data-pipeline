```python
from datetime import datetime
import pandas as pd

def extract_news(news_items):
    """
    Pull publish time, title, and summary from a yfinance news list.

    Parameters
    ----------
    news_items : list of dict
        Output of Ticker.news.

    Returns
    -------
    pd.DataFrame
        Columns: time, title, summary.
    """
    rows = []
    for item in news_items:
        # Newer yfinance nests most fields under 'content'
        content = item.get('content', item)

        title = content.get('title')
        summary = content.get('summary')

        timestamp = item.get('providerPublishTime') or content.get('pubDate')
        if isinstance(timestamp, (int, float)):
            time = datetime.fromtimestamp(timestamp)
        else:
            time = timestamp  # already a string in some versions

        rows.append({'time': time, 'title': title, 'summary': summary})

    return pd.DataFrame(rows)

news_df = extract_news(stock.news)
```
